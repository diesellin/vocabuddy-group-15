# 英文單字學習清單

這裡新增了三個英文單字及其中文解釋。

```python
# 定義英文單字與中文解釋的字典
vocabulary_list = {
    "ephemeral": "短暫的，瞬息即逝的",
    "resilient": "有彈性的，能迅速恢復的",
    "scrutinize": "詳細檢查，細看"
}

# 顯示單字與解釋
for word, definition in vocabulary_list.items():
    print(f"{word}: {definition}")
```
