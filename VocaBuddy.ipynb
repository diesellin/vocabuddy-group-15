```python
import json
from IPython.display import HTML, display, Javascript, clear_output

# Prepare the vocabulary list in the format expected by the JavaScript
js_words_array = [[english, chinese] for english, chinese in vocabulary_list.items()]
words_json_for_js = json.dumps(js_words_array, ensure_ascii=False)

# HTML part (without the script block)
html_part = f"""
<!DOCTYPE html>
<html lang="zh-Hant">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>YTP VocaBuddy</title>
<style>
  body{{font-family:system-ui,sans-serif;background:#f4f6f5;margin:0;padding:24px;text-align:center}}
  h1{{color:#0f766e}}
  #card{{max-width:340px;margin:32px auto;padding:40px 24px;background:#fff;border-radius:14px;
        box-shadow:0 4px 16px rgba(0,0,0,.08);font-size:2rem;cursor:pointer;user-select:none}}
  #hint{{color:#888;font-size:.9rem}}
  button{{margin-top:16px;padding:10px 22px;font-size:1rem;border:none;border-radius:8px;
         background:#0f766e;color:#fff;cursor:pointer}}
</style>
</head>
<body>
<h1>YTP VocaBuddy 背單字夥伴</h1>
<div id="card">algorithm</div>
<p id="hint">點卡片看中文，點「下一張」換單字</p>
<button id="nextButton" type="button">下一張</button>
</body>
</html>
"""

# JavaScript part (extracted from HTML and wrapped in display(Javascript))
js_content = f"""
const WORDS = {words_json_for_js};
let i = 0, flipped = false;
const card = document.getElementById("card");
const nextButton = document.getElementById("nextButton");

// Add a check to ensure elements are available before binding events
if (card && nextButton) {{
    card.onclick = () => {{ flipped = {"!"}flipped; render(); }};
    nextButton.onclick = () => {{ i = (i + 1) % WORDS.length; flipped = false; render(); }};

    function render(){{
        if (WORDS.length > 0) {{
            card.textContent = WORDS[i][flipped ? 1 : 0];
        }} else {{
            card.textContent = "No words loaded";
        }}
    }}
    // Initial render
    render();
}} else {{
    console.error("VocaBuddy: Required DOM elements (card or nextButton) not found.");
}}
"""

clear_output(wait=True) # Clear previous output to show the new UI
display(HTML(html_part))
display(Javascript(js_content))
```

```python
import json
import random
from IPython.display import HTML, display, Javascript, clear_output

# Safely get vocabulary_list or define a default one if it doesn't exist
try:
    vocabulary_list
except NameError:
    vocabulary_list = {
        'ephemeral': '短暫的，瞬息即逝的',
        'resilient': '有彈性的，能迅速恢復的',
        'scrutinize': '詳細檢查，細看',
        'ubiquitous': '無處不在的，普遍存在的',
        'benevolent': '仁慈的，善意的',
        'cacophony': '刺耳的聲音，不和諧的聲音',
        'gregarious': '愛交際的，群居的',
        'abc': '前三個英文字母'
    }

# Prepare vocabulary list
js_words_array = [[english, chinese] for english, chinese in vocabulary_list.items()]
words_json_for_js = json.dumps(js_words_array, ensure_ascii=False)

html_quiz = """
<!DOCTYPE html>
<html lang="zh-Hant">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>YTP VocaBuddy - Quiz</title>
<style>
  body { font-family: system-ui, sans-serif; background: #f4f6f5; margin: 0; padding: 24px; text-align: center; }
  h1 { color: #0f766e; }
  #quiz-container { max-width: 450px; margin: 32px auto; padding: 24px; background: #fff; border-radius: 14px; box-shadow: 0 4px 16px rgba(0,0,0,.08); }
  #question-word { font-size: 2.2rem; color: #1e293b; margin-bottom: 20px; font-weight: bold; }
  .option-btn { display: block; width: 100%; margin: 10px 0; padding: 12px; font-size: 1.1rem; border: 2px solid #e2e8f0; border-radius: 8px; background: #fff; cursor: pointer; text-align: left; transition: all 0.2s; }
  .option-btn:hover { background: #f1f5f9; border-color: #cbd5e1; }
  #score-board { margin-top: 20px; font-size: 1rem; color: #64748b; }
  #feedback { margin-top: 15px; font-size: 1.2rem; font-weight: bold; height: 30px; }
  button#nextQuizButton { margin-top: 16px; padding: 10px 22px; font-size: 1rem; border: none; border-radius: 8px; background: #0f766e; color: #fff; cursor: pointer; display: none; margin-left: auto; margin-right: auto; }
</style>
</head>
<body>
<h1>YTP VocaBuddy 單字挑戰賽</h1>
<div id="quiz-container">
  <div id="question-word">Loading...</div>
  <div id="options-container"></div>
  <div id="feedback"></div>
  <button id="nextQuizButton" type="button">下一題</button>
  <div id="score-board">得分: <span id="score">0</span> / <span id="total">0</span></div>
</div>
</body>
</html>
"""

js_quiz = f"""
const WORDS = {words_json_for_js};
let score = 0;
let total = 0;
let currentQuestion = null;
let answered = false;

const wordEl = document.getElementById("question-word");
const optionsContainer = document.getElementById("options-container");
const feedbackEl = document.getElementById("feedback");
const nextBtn = document.getElementById("nextQuizButton");
const scoreEl = document.getElementById("score");
const totalEl = document.getElementById("total");

function getRandomOptions(correctAnswer, allWords) {{    let options = [correctAnswer];
    let remaining = allWords.filter(w => w[1] !== correctAnswer);

    // Shuffle remaining and pick 3
    remaining.sort(() => 0.5 - Math.random());
    for (let i = 0; i < Math.min(3, remaining.length); i++) {{
        options.push(remaining[i][1]);
    }}

    // Shuffle the final 4 options
    return options.sort(() => 0.5 - Math.random());
}}

function generateQuestion() {{    if (WORDS.length < 4) {{
        wordEl.textContent = "單字量不足（至少需要 4 個單字）";
        return;
    }}

    answered = false;
    feedbackEl.textContent = "";
    feedbackEl.style.color = "black";
    nextBtn.style.display = "none";

    // Pick a random word
    const randomIndex = Math.floor(Math.random() * WORDS.length);
    currentQuestion = WORDS[randomIndex];

    wordEl.textContent = currentQuestion[0];
    const options = getRandomOptions(currentQuestion[1], WORDS);

    optionsContainer.innerHTML = "";
    options.forEach(opt => {{
        const btn = document.createElement("button");
        btn.className = "option-btn";
        btn.textContent = opt;
        btn.onclick = () => selectOption(btn, opt);
        optionsContainer.appendChild(btn);
    }});
}}

function selectOption(selectedBtn, selectedValue) {{    if (answered) return;
    answered = true;
    total++;

    const correctValue = currentQuestion[1];
    const buttons = optionsContainer.getElementsByClassName("option-btn");

    for (let btn of buttons) {{
        btn.style.cursor = "default";
        if (btn.textContent === correctValue) {{
            btn.style.background = "#d1fae5";
            btn.style.borderColor = "#10b981";
            btn.style.color = "#065f46";
        }} else if (btn === selectedBtn) {{
            btn.style.background = "#fee2e2";
            btn.style.borderColor = "#ef4444";
            btn.style.color = "#991b1b";
        }}
    }}

    if (selectedValue === correctValue) {{
        score++;
        feedbackEl.textContent = "答對了！🎉";
        feedbackEl.style.color = "#10b981";
    }} else {{
        feedbackEl.textContent = `答錯了，答案是: ${{correctValue}}`;
        feedbackEl.style.color = "#ef4444";
    }}

    scoreEl.textContent = score;
    totalEl.textContent = total;
    nextBtn.style.display = "block";
}}

nextBtn.onclick = generateQuestion;

// Start first question
generateQuestion();
"""

clear_output(wait=True)
display(HTML(html_quiz))
display(Javascript(js_quiz))
```
